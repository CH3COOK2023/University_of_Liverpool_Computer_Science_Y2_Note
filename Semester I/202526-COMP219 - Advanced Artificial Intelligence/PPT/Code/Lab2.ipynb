### Getting Set Up and Basic Visualisation
In Lab 1, we worked with the iris and the digits datasets. We will stick with the same datasets this week; however, we will take a different approach when setting things up.

In this lab, rather than loading the dataset using a built-in library function, we will import the data from our local machine. Let's import and store the dataset by using the Pandas function "read_csv()". We will also need to create indexing for later use.

```python
#import iris.csv
iris = pd.read_csv("iris.csv")

#add indexing
#id becomes the first column: we do this using the insert function at index 0
iris.insert(0, "id", range(0, len(iris)))
```

Now that we have dataset imported and stored, we can start working. Let's have a quick look at what we are working with in a different and perhaps slightly less convoluted way than we did last week. It is always a good practice to inspect the raw data.

```python
#Quick overview
print(iris.head())
print("Size of the dataset: ", iris.size)
print(iris.shape)
```

Let's visualise the data next. We can call the plot() function (part of Pandas) on the iris dataset. On the x-axis, we can plot sepal length, and on the y-axis, we can plot sepal width. The code below will only execute if you have imported and stored the iris dataset under the variable name "iris".

```python
#Scatterplot
iris.plot(kind = "scatter", x = "sepal_length", y = "sepal_width")
plt.grid() #Add a grid to our plot
```

#### The Seaborn Library
Seaborn includes various visualisation tools that allow us to customise our plots more freely. The above plot is not as informative as we would like it to be; with Seaborn, we can easily add additional labels and colour distinctions.

With Seaborn installed, let's improve our plot by adding appropriate colour coding and additional labels. The plot below will now have more information.

```python
#Improved scatterplot
sns.FacetGrid(iris, hue = "species") \
.map(plt.scatter, "sepal_length", "sepal_width") \
.add_legend()
```

We will now have a look at the following visualisation techniques:
- Boxplots: these allow us to investigate individual features.
- Violin plots: these are similar to boxplots, however, we can get a better view of density/sparsity of data.
- Kernel Density Estimation plots: these plots allow us to visualise the probability density of continuous variables.
- Pairplots: useful for investigating bivariate relations between pairs of features.

Observe the following blocks of code and their respective outputs as you will need them later:

```python
#Boxplots
sns.boxplot(x="species", y="petal_length", data=iris)
ax = sns.boxplot(x="species", y="petal_length", data=iris)
ax = sns.stripplot(x="species", y="petal_length", data=iris, jitter=True, edgecolor="gray")
```

```python
#Violin plots
sns.violinplot(x="species", y="petal_length", data=iris)
```

```python
#KDE
sns.FacetGrid(iris, hue="species") \
   .map(sns.kdeplot, "petal_length") \
   .add_legend()
```

```python
#Pairplots
sns.pairplot(iris.drop("id", axis=1), hue="species", size=3)
```
