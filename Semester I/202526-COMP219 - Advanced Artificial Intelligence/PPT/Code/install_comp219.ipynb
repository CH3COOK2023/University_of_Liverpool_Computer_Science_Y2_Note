```python
import importlib
import subprocess
import sys
import platform

packages = [
    "pandas", "numpy", "matplotlib", "tensorflow",
    "scikit-learn", "torch", "torchvision"
]

missing = [pkg for pkg in packages if importlib.util.find_spec(pkg) is None]

if missing:
    print(f"Missing packages: {', '.join(missing)}")

    #use conda if available, if not, use pip
    try:
        if platform.system() == "Windows":
            subprocess.run("conda install -y " + " ".join(missing), shell=True, check=True)
        else:
            subprocess.run(["conda", "install", "-y"] + missing, check=True)
        print("new packages installed. please restart your kernel.")
    except subprocess.CalledProcessError:
        print("conda failed, using pip...")
        subprocess.run([sys.executable, "-m", "pip", "install"] + missing, check=True)
        print("installed missing packages using pip. please restart your kernel")
else:
    print("all required packages are already installed")

print(sys.executable)
#!which python #linux/mac 
!where python #windows
!conda list | findstr /R "pandas numpy matplotlib tensorflow scikit-learn torch torchvision"
```
