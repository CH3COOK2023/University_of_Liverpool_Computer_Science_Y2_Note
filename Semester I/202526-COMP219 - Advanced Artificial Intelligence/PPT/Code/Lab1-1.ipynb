### Task 2 - Working with datasets
As part of the course, you will be required to import datasets. Some datasets are available as part of some Python packages. The scikit-learn package has a nice introductory dataset called the iris dataset (if you wish to know more, please see [this link](https://en.wikipedia.org/wiki/Iris_flower_data_set)).

For this task, you don't need to write any code; simply run each chunk and try to understand the output.

```python
#First, we import our datasets
from sklearn import datasets

#Load the iris dataset
iris_data = datasets.load_iris()

#Let's observe the characteristics of the dataset.
iris_data
```

With all that set, we can now do some basic operations on the dataset.

If we scroll down, we can see 'target_names' and 'feature_names' - we need to assign these to their respective variables. Once we have done that, we can draw a sample from our dataset and present it in a meaningful way. Note that in this context, feature represents some physical description of a flower while target describes the species. The variable 'data' stores an array of values for each record.

```python
import random

#Assign feature and target names, and data
features = iris_data.feature_names
targets = iris_data.target_names
y = iris_data.target
x = iris_data.data

#Print features and targets
print("Feature names: ", features)
print("Target names: ", targets)

#Print the number of rows (there are several ways of doing this, some are more efficient than others)
#In this example, we'll use shape[] where 0 = number of rows (records or y) and 1 = number of columns (features or x)
dataset_length = y.shape[0]
print("Number of records: ", dataset_length)

#Look at the first 10 entries
for i in range (10):
    print(f"Sample {i + 1}: {x[i]} (Class: {y[i]}, Species: {targets[y[i]]})")
```
